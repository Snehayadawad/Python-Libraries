```python
employees = {}
while True:
  name = input("Enter employee name(or'quit):")
  if name == "quit":
    break
  dept = input("Enter department:")
  employees[name] = dept
print("\nEmployee Directory")
for name, dept in employees.items():
  print(f"Employee:{name}|Department:{dept}")
```
